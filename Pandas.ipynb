# Duplicates

```python
mtcars = data('mtcars')
mtcars.head(2)
```

```python
mtcars2 = mtcars.copy(). reset_index().rename(columns={'index':'car'})
mtcars2.head()
```

```python
mtcars2.shape
```

```python
mtcars2[mtcars2['car'] == 'Mazda RX4']
```

```python
mtcars2[mtcars2['cyl'] == 'Hornet 4 Drive'] #why the same data not showing for 'cyl'?
```

```python
mazda2 = mtcars2[mtcars2['car'] == 'Mazda RX4']
datson2 = mtcars2[mtcars2['car'] == 'Datsun 710']
```

```python
mtcars_dup = pd.concat([mtcars2, mazda2, datson2, mazda2], ignore_index=True)
mtcars_dup
```
