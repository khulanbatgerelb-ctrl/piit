```python
import pandas as pd
import numpy as np
```

```python
data = [
    [1, 'Khulan', 78, 90],
    [2, 'Batgerel', 80, 90],
    [3, 'Dhiraj', 78, 90],
    [4, 'Manisha', 88, 76],
    [5, 'Aelita', 56, 90],
]
df = pd.DataFrame(data, columns=['Rollno', 'Name', 'Maths', 'English'])
df
```

# wide long

```python
df_long = df.melt(id_vars=['Rollno', 'Name'],
                 value_vars=['Maths', 'English'],
                 var_name='Subject',
                 value_name='Marks')
df_long
```

```python
wide_df = df_long.pivot(index=['Rollno', 'Name'], columns='Subject',
                       values='Marks').reset_index(drop=False).rename_axis(None, axis=1)
wide_df
```

```python
wide_df2 = df_long.pivot_table(index=['Rollno','Name'], columns='Subject',
                               values='Marks', aggfunc='sum').reset_index(drop=False).rename_axis(None, axis=1)
wide_df2
```

# Group By

```python
df_long.groupby('Subject')['Marks'].agg(['sum', 'mean', 'count'])
```

# Missing Values

```python
import numpy as np
import pandas as pd
data= [
    [1, 'Khulan', None, 90],
    [2, 'Bat', 90,90],
    [3, 'Dhiraj', 80,90],
    [4, 'Manisha', np.nan, 70],
    [5, 'Gerel', np.nan, None],
]
df = pd.DataFrame(data, columns=['Rollno', 'Name', 'Maths','English' ])
df
```

```python
df.isna().sum()
```

```python
df.isna().sum(axis=1)
```

```python
df.fillna(50)
```

```python
df['Maths'].mean()
```

```python
df['Maths'] = df['Maths'].fillna(df['Maths'].mean())
df
#filling data with 85.0
```

```python
df['English'] = df['English'].fillna(df['English'].mean())
df
```
