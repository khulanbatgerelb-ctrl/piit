```python
import numpy as np
import pandas as pd
```

```python
pd.set_option('display.max_columns',None)
```

```python
from pydataset import data
mtcars = data('mtcars')
mtcars.head()
```

```python
mtcars.dtypes
```

```python
mtcars.head(3)
```

```python
mtcars = data('mtcars')
mtcars = mtcars.astype({ 'cyl': 'category', 'vs': 'category', 'carb': 'category', 'am': 'category', 'gear': 'category'})
```

```python
mtcars.dtypes
```

```python
mtcars.head()
```

```python
mtcars['am'] =mtcars['am'].map({0: 'auto', 1:'manual'})
mtcars.head() #sorting by manual or automatic engine
```

```python
mtcars['mpg'].max()
```

```python
bins = [0, 17, 29, mtcars['mpg'].max()]
labels = ['Low', 'Medium', 'High']
```

```python
mtcars['mpg_level'] = pd.cut(mtcars['mpg'], bins=bins, labels=labels, include_lowest=True)
```

```python
mtcars.head()
```

```python
mtcars.groupby('mpg_level')['mpg'].count()
```

```python
mtcars.mpg_level.value_counts()
```

```python
mtcars['vs'] = mtcars['vs'].map({0: 'Vshaped', 1:'Straight'})
mtcars.head()
```

```python
mtcars.reset_index()
```

```python
mtcars.index
```

```python
mtcars.head()
```

```python
mtcars['car_name'] = mtcars.index
mtcars.head()
```

```python
mtcars['model'] = mtcars.index #carnames & model & carno.
```

```python
mtcars.head()
```

```python
mtcars.dtypes
```

```python
print(len(mtcars[mtcars['mpg'] > 25]))
```

```python
mtcars[mtcars['mpg'] > 25]
```

```python
mtcars[mtcars['wt'] >= 5]
```

```python
mtcars[mtcars['mpg_level'] == 'High' ]
```

```python
mtcars.query("mpg < 23")
```

```python
mtcars.filter(items = ['mpg','hp'])
```

```python
mtcars[mtcars['am'] == 'auto']
```

```python
mtcars[(mtcars['am'] == 'auto') & (mtcars['cyl'] != '5') & (mtcars['mpg'] > 22 )]
# <> 5, mpg>22
```

```python
mtcars[mtcars['mpg'].between(left=23, right=26.1, inclusive='left')]
#between condition 23-26.1
#inclusive {both, neither, left, right}
```

```python
mtcars[mtcars['mpg'].between(23, 30, inclusive= 'neither')]
```

```python
mtcars[mtcars['gear'].isin ([4,5])]
#look for data within 4,5 values in gear
```

```python
mtcars[mtcars['cyl'].isin ([4,6])]
#Cyl value including 4,6
```

```python
mtcars[ (mtcars['gear'].isin([4,5])) & (mtcars['am'] == 'auto')]
# two different condition with different data values using & AND
```

```python
mtcars[~mtcars['gear'].isin([4,5])]
```

```python
import pandas as pd
import numpy as np
from pydataset import data
mtcars = data('mtcars')
mtcars.shape
```

```python
mtcars.pivot_table(values = 'mpg', index='gear', observed=False)
```

```python
mtcars.pivot_table(values = 'hp', index='cyl', observed=False)
```

```python
pd.pivot_table( data=mtcars, observed=False,
              index='gear',
              values='mpg',
              aggfunc='median'
)
```

```python
pivot1 = pd.pivot_table(
    data=mtcars, observed=False,
    index='gear',
    columns='am',
    values='mpg',
    aggfunc='mean'
)
```

```python
pivot1
```

```python
mtcars.am.value_counts()
mtcars.cyl.value_counts()
```

```python
pd.pivot_table ( observed=False, data=mtcars, index='gear', columns='am', values='mpg', \
               aggfunc=['mean', 'max', 'min', 'count', 'sum'] ).reset_index()
```

```python
pd.pivot_table(dat= mtcars, index=['gear','am'], columns='cyl', values='mpg', \
              aggfucn=['mean', 'max', 'min' , 'count', 'sum', 'std'] ).reset_index()
```

```python
pd.pivot_table(data=mtcars, index='cyl', columns='carb', values='hp', aggfunc=['min','max','count'])
```
