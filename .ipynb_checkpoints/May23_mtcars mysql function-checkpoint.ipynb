```python
# use piit;
select * from mtcars;
select gear, mpg, row_number() over (partition by gear order by mpg desc) as row_num from mtcars;
select gear, mpg, row_number() over (partition by gear order by mpg asc) as row_num from mtcars;
select gear, mpg, row_number() over ( order by mpg asc) as row_num from mtcars;
select mpg, cyl, rank() over (order by mpg desc) as mpg_rank from mtcars;
select am, hp , rank() over (order by hp desc) as hp_rank from mtcars;
select am, hp, rank() over (partition by am order by hp desc) as hp_rank from mtcars;
select am, hp,dense_rank() over (partition by am order by hp asc) as hp_dense_rank from mtcars;
select am, hp, rank() over (partition by am order by hp asc) as hp_rank from mtcars;
select am, hp, rank() over (partition by am order by hp asc) as hp_rank from mtcars order by am desc;
select * from mtcars;
select cyl, am, sum(mpg) over (order by mpg) as running_total_mpg  from mtcars;
select cyl, am, mpg, sum(mpg) over (order by mpg) as running_total_mpg from mtcars;
select cyl, am, mpg, round(sum(mpg) over (order by mpg),1) as running_total_mpg from mtcars;
select cyl, am, mpg, lead(mpg) over (order by mpg) as running_total_mpg from mtcars;
select cyl, am, mpg, lead(mpg) over (order by mpg) as next_mpg from mtcars;
select cyl, am, mpg, lag(mpg) over (order by mpg) as previous_mpg from mtcars;
select curdate() from mtcars;
select cyl, mpg, am, curdate() from mtcars;
select cyl, mpg, am, now() from mtcars
select cyl, mpg, am, curdate(), date_format(curdate(), '%d-%m-%y') as formatted_date  from mtcars;
SELECT e.empno, e.ename, e.job, e.sal, d.dname, d.loc
    -> FROM emp e
    -> INNER JOIN dept d ON e.deptno = d.deptno;
select e.empno, e.ename, e.deptno, d.dname from emp e, dept d where e.deptno = d.deptno;
select e.empno, e.ename, e.deptno, d.dname from emp e inner join dept d on e.deptno = d.deptno;
```
