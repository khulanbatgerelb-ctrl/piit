# SQL Python connect

```python
from sqlalchemy import create_engine, text
import pymysql
import pandas as pd
import os
```

```python
engine1 = create_engine("mysql+pymysql://root:piit2025@localhost:3306/piit")
engine1
```

```python
# function to run the query with parameters or query
def q(sql, **params):
    with engine1.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

```python
# Total compensation sal+comm
empdept['total_comp'] = empdept['sal'] + empdept['comm']
empdept[['ename','sal','comm','total_comp']].sort_values('total_comp', ascending=False)
```

```python
# Sequencing order number
result = empdept[['ename','sal','comm','total_comp']]\
         .sort_values('total_comp', ascending=False)\
         .reset_index(drop=True)

result.index += 1  # start from 1 instead of 0
result
```
