# Graphs with Matplotlib Library

```python
import pandas as pd
import numpy as np
from pydataset import data
```

```python
df = data ('mtcars')
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
# 1. Histogram
plt.figure
plt.hist(df['mpg'])
plt.title("Distribution of MPG")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.show()
```

```python
# 1.1 Cyl Histogram
plt.figure
plt.hist(df['cyl'])
plt.title("Distribution of CYL")
plt.xlabel("CYL")
plt.ylabel("Frequency")
plt.show()
```

```python
1.2 #DISP Histogram
plt.figure
plt.hist(df['disp'])
plt.title("GRAPH of DISP")
plt.xlabel("DISP")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2. Scatter
plt.figure
plt.scatter(df['hp'], df['mpg'])
plt.title("HP vs MPG")
plt.xlabel("Horsepower")
plt.ylabel("MPG")
plt.show()
```

```python
# 2. Scatter for WT vs QSEC
plt.figure
plt.scatter(df['wt'], df['qsec'])
plt.title("WT vs QSEC")
plt.xlabel("WorkTruck")
plt.ylabel("Quarter-Mile Time")
plt.show()
```

```python
#3. Boxplot
plt.figure()
df.boxplot(column='mpg', by='cyl')
plt.title ("MPG by Cylinders")
plt.suptitle("1.0")
plt.show()
```
