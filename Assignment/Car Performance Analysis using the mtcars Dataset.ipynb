```python
#car performance, fuel efficiency, and engine features
performance = mtcars[['hp', 'qsec', 'wt']]
fuel_efficiency = mtcars[['mpg']]
engine_features = mtcars[['cyl', 'disp', 'drat', 'carb']]
```

```python
#visualization data in table
performance.head()
fuel_efficiency.head()
engine_features.head()
```

```python
# Average MPG
mtcars['mpg'].mean()

# Top 5 fuel-efficient cars
mtcars.sort_values(by='mpg', ascending=False).head()

# Average horsepower by cylinders
mtcars.groupby('cyl')['hp'].mean()
```

```python
# Correlation
mtcars.corr()

# Relationship between HP and MPG
mtcars[['hp', 'mpg']].corr()

# Grouped insights
mtcars.groupby('cyl').agg({
    'mpg': 'mean',
    'hp': 'mean',
    'wt': 'mean'
})
```

```python
# Define categories FIRST
performance = mtcars[['hp', 'qsec', 'wt']]
fuel_efficiency = mtcars[['mpg']]
engine_features = mtcars[['cyl', 'disp', 'drat', 'carb']]

# Then print
print("\nPerformance:\n", performance.head())
print("\nFuel Efficiency:\n", fuel_efficiency.head())
print("\nEngine Features:\n", engine_features.head())
```

```python
# using \n for naming the title 
print("\nPerformance: \n", performance.head())
print("\nFuel Efficiency:\n", fuel_efficiency.head())
print("\nEngine Features:\n", engine_features.head())
```

```python
# Scatter chart for Fuel Efficiency vs Horsepower
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=mtcars, x='hp', y='mpg')
plt.title("Fuel Efficiency vs Horsepower")
plt.show()
```

```python
# Scatter chart for Engine displacement vs MPG
sns.scatterplot(data=mtcars, x='disp', y='mpg')
plt.title("Engine Displacement vs MPG")
plt.show()
```

```python
summary = pd.concat([
    performance.mean().rename("Performance Avg"),
    fuel_efficiency.mean().rename("Fuel Avg"),
    engine_features.mean().rename("Engine Avg")
], axis=1)

print(summary)
```

```python
# histogram
mtcars ['mpg'].hist()
plt.title("MPG distribution")
plt.show()
```

```python
# bar chart
mtcars.groupby('cyl')['mpg'].mean().plot(kind='bar')
plt.title("Average MPG by Cylinders")
plt.show()
```

```python
# Cars with higher horsepower tend to have lower MPG
# 4-cylinder cars are more fuel efficient than 6 or 8-cylinder
# Heavier cars (weight) reduce fuel efficiency
# Strong negative correlation between hp and mpg
```
