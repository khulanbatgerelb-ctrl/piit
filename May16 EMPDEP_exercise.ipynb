# SQL Python connect

```python
from sqlalchemy import create_engine, text
import pymysql
import pandas as pd
import os
```

```python
engine1 = create_engine("mysql+pymysql://root:piit2025@localhost:3306/piit")
engine1
```

```python
# function to run the query with parameters or query
def q(sql, **params):
    with engine1.connect() as conn:
        return pd.read_sql(text(sql), conn, params=params)
```

```python
q('show tables')
```

```python
q('select * from emp')
```

```python
q('select * from dept')
```

```python
# List the names of analysts and salesmen
q('select ename, job from emp where job in ("Analyst","Salesman")')
```

```python
# List details of employees who have joined before 30 Sep 81.
q('select ename, hiredate from emp where hiredate < "1981-09-30"')
```

```python
# List names of employees who are not managers. (! command = not equal )
q('select ename, job from emp where job != "MANAGER"')
```

```python
# List the names of employees whose employee numbers are 7369, 7521, 7839, 7934, 7788.
# using IN command = is same 
q('select ename, empno from emp where empno IN (7369,7521, 7839, 7934, 7788)')
```

```python
# List employees not belonging to department 30, 40, or 10.
q('select ename, deptno from emp where deptno not in (30,40,10)')
```

```python
# List employee names for those who have joined between 30 June and 31 Dec. ‘81.
q('select ename, hiredate from emp where hiredate between "1981-06-30" and "1981-12-31"')
```

```python
# List the different designations in the company.
q('select distinct job from emp')
```

```python
#List the names of employees who are not eligible for commission.
q('select ename, comm from emp where comm is null or comm = 0')
```

```python
# List the name and designation of the employee who does not report to anybody.
q('select ename, job from emp where job = "president"')
```

```python
# List the employees not assigned to any department.
q('select ename, deptno from emp where deptno = null')
```

```python
# List the employees who are eligible for commission.
q('select ename, comm from emp where comm is not null or comm > 0')
```

```python
# List employees whose names either start or end with “S”.
# using % starting with %S or ending with S% using LIKE command
q('select ename from emp where ename like "S%" or ename like "%S"')
```

```python
# List names of employees whose names have “i” as the second character.
# secoend means "_"
q('select ename from emp where ename like "_i%"')
```

```python
# 3rd character "__"
q('select ename from emp where ename like "__i%"')
```

```python
# List the number of employees working with the company.
q('select count(*) as total_employees from emp')
```

```python
# List the number of designations available in the EMP table.
q('select count(*) as total_deptno from emp')
```

```python
# List the total salaries paid to the employees.
q('select sum(sal) from emp')
```

```python
# List the maximum, minimum and average salary in the company.
q('select max(sal) as max_salary, min(sal) as min_salary, avg(sal) as avg_salary from emp')
```

```python
# List the maximum salary paid to a salesman.
q('select max(sal) as max_salary from emp where job = "Salesman"')
```
